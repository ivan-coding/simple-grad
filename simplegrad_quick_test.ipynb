# Micrograd quick test

Build a small expression, run backpropagation and inspect the computation graph.

```python
from simplegrad import Value
from simplegrad.visualize import build_graph

a = Value(0.4, label="a")
b = Value(0.5, label="b")
c = Value(-0.4, label="c")
d = Value(0.2, label="d")
result = (a * b + a * c + d).tanh()
result.label = "result"
result.backward()
```

```python
build_graph(result)
```
